### Understanding your data

When getting dataset ask 7 questions

1) How big is data 
Use .shape and chunksize if large

2) How data look like
Use .head or .sample

3) Data type of columns
Use .info and change columns data type if needed

4) Are there any missing values
Use .isnull().sum() to get number of missing values and can be filled or dropped

5) How data look mathematically
Use .describe() to get datas like mean max min var and many more

6) Are there any duplicate values?
Use .duplicated().sum() to get number of duplicate values and then drop those rows

7) Correlation between columns
Basically identifying the corelation between columns to know whether they change the output or not and whether they can be merged or not and drop not needed columns
